```python
#titanic dataset
df=pd.read_csv('/content/titanic.csv')
df['Name']
```

```python
#common functions
#df['Name'].str.lower()
df['Name'].str.upper()
df['Name'].str.capitalize() #first letter capital
df['Name'].str.title() #first letter capital of each word in the name
```

```python
#print the name of the person with maximum letter
df['Name'][df['Name'].str.len()==82]
```

```python
#strip - removes leading and trailing spaces
df['Name'].str.strip()
```

```python
#split
df['Name'].str.split(',')
```

```python
#split -> get = getting the first element from the list
df['lastname']=df['Name'].str.split(',').str.get(0)
```

```python
df['lastname']
```

```python
df.head()
```

```python
df['Name']
```

```python
#getting the name
df['FirstName']=df['Name'].str.split(',').str.get(1).str.split('.').str.get(1)
```

```python
df
```

```python
#replace
df['Embarked']=df['Embarked'].str.replace('S','T')
df
```

```python
#filtering
df[df['FirstName'].str.strip().str.startswith('A')]
df[df['FirstName'].str.strip().str.endswith('A')]
#isdigit
df[df['FirstName'].str.isdigit()]
```

```python
#applying regular expression
df[df['FirstName'].str.contains('John',case=False)]
```

```python
#slicing
df['Name'].str[:4]
df['Name'].str[::2]
```
