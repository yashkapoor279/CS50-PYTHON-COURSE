DATE RANGE FUNCTION

```python
#generates daily dates in a given range
pd.date_range(start='2026/7/3',end='2026/9/13',freq='D')
pd.date_range(start='2026/7/3',end='2026/9/13',freq='2D') #alternate dates
pd.date_range(start='2026/7/3',end='2026/9/13',freq='3D') #after 3 days
```

```python
#business days
pd.date_range(start='2026/7/3',end='2026/9/13',freq='B')
```

```python
#one week per day
pd.date_range(start='2026/7/3',end='2026/9/13',freq='W-TUE')
```

```python
#hourly data
pd.date_range(start='2026/7/3',end='2026/9/13',freq='H')
```

```python
pd.date_range(start='2026/7/3',end='2026/9/13',freq='23H')
```

```python
#Month end
pd.date_range(start='2026/7/3',end='2026/9/13',freq='M')
```

```python
#month start
pd.date_range(start='2026/7/3',end='2026/9/13',freq='MS')
```

```python
#A->YEAR END
```

```python
pd.date_range(start='2026/7/3',periods=25,freq='D') #aaj se 25 din tk
```
