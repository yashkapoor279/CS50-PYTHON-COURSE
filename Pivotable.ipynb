```python
import pandas as pd
import numpy as np
```

```python
#pivot table takes simple column wise data as input and groups the entries into 2D table
```

```python
import seaborn as sns
```

```python
df=sns.load_dataset('tips') #random default data in sns library
df.head()
```

```python
df.groupby('sex')['total_bill'].mean()
```

```python
df.groupby(['sex','smoker'])['total_bill'].mean().unstack()
```

```python
#pivot table -(shortcut way) takes simple col data as input and groups the entries into 2D data
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='std') #by default it calculates the mean
```

```python
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill',aggfunc='sum',margins=True)
```

```python
#plotting graph
df = pd.read_csv('/content/expense_data.csv')
```

```python
df['Category'].value_counts()
```

```python
df.info()
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df.info()
```

```python
df['month']=df['Date'].dt.month_name()
```

```python
df.head()
```

```python
df.pivot_table(index='month',columns='Category',values='INR',aggfunc='sum',fill_value=0).plot()
```

```python
df.pivot_table(index='month',columns='Account',values='INR',aggfunc='sum',fill_value=0)
```

```python
df.pivot_table(index='month',columns='Account',values='INR',aggfunc='sum',fill_value=0).plot()
```
