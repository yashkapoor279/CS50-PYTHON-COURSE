```python
batters=pd.read_csv('batter.csv')
sample_df=batters.sample(25)
```

```python
plt.figure(figsize=[12,10])
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color="black")
plt.axvline(20,color="black")

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x=[1,2,3,4]
y=[5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 4')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 2')
plt.tight_layout()
```

```python
fig,ax=plt.subplots(figsize=[15,6])

ax.scatter(batters['avg'],batters['strike_rate'],1)
ax.scatter(batters['runs'],batters['avg'])
ax.set_title('IPL DATA')
ax.set_title('naukri')
```

```python
fig,ax = plt.subplots(2,2)
ax[0,0].scatter(batters['avg'],batters['strike_rate'])
ax[0,0].set_title('Graph1')
ax[0,1].scatter(batters['runs'],batters['avg'],color='red')
ax[0,1].set_title('Graph2')
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])


```

```python
fig = plt.figure()


ax = plt.subplot(projection = '3d')

ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'],color = 'red')
ax.set_title('IPL BATSMAN')
ax.set_xlabel('runs')
ax.set_ylabel('average')
ax.set_zlabel('strike_rate')
```

```python
#3D LINE PLOT
x=[0,1,5]
y=[0,10,13]
z=[0,13,20]
ax=plt.subplot(projection = '3d')
ax.scatter3D(x,y,z)
ax.plot3D(x,y,z,color='red')
```

```python
#3D SURFACE PLOT
x= np.linspace(-10,10,100)
y= np.linspace(-10,10,100)
```

```python
xx,yy = np.meshgrid(x,y)
```

```python
z = xx**2+yy**2
```

```python
z.shape
```

```python
fig=plt.figure(figsize=[12,8])
ax=plt.subplot(projection ='3d')
ax.plot_surface(xx,yy,z)
```

```python
fig=plt.figure(figsize=[12,8])
ax=plt.subplot(projection ='3d')
ax.plot_surface(xx,yy,z)
```

```python
#contourplot = represent 3d graph in 2d
fig=plt.figure(figsize=[12,8])
ax=plt.subplot()
# ax.contour(xx,yy,z)
ax.contourf(xx,yy,z)
```
