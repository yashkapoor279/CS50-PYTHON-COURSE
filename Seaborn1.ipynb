```python
#SEABORN
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
#Relational Plots - statistical relation
tips=sns.load_dataset('tips')
tips
```

```python
#scatter plot = axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#relplot -> figure level ->square shape
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',kind='line',hue='sex')
```

```python
#hue parameter
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='time',style='sex',size='size')
```

```python
#line plot
gap = px.data.gapminder()
```

```python
temp_df=gap[gap['country']=='India']
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp') #axes plot
```

```python
#using relplot
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line',hue='country')
```

```python
temp_df1=gap[gap['country'].isin(['India','Brazil','Germany','China'])]
```

```python
temp_df1
```

```python
#figure
sns.relplot(data=temp_df1,x='year',y='lifeExp',kind='line',hue='country',style='continent',size='continent')
```

```python
#axes level
sns.lineplot(data=temp_df1,x='year',y='lifeExp',hue='country')
```

```python
#facetplot - different plots for categorical columns -> can only be made of figure level func
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
#col wrap -> used to wrp in rows and columns
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap = 4)
```

```python
#distribution plot - univariate analysis
#histplot
sns.histplot(data=tips,x='total_bill')
```

```python
#figure level = displot
#axes level -> histplot->kde plot->rugplot
```

```python
sns.displot(data=tips,x='total_bill',kind='hist',bins=42)
```

```python
sns.displot(data=tips,x='day',kind='hist')
```

```python
#hue parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
titanic=sns.load_dataset('titanic')
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex')
```

```python
sns.displot(data=titanic,x='age',kind='hist',col='sex')
```

```python
#KDE PLOT - KERNEL DENSITY ESTIMATION
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='day')
```

```python
#rug plot - plot marginal distribution by drawing ticks along x and y axis
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
#bivarateanalysis using histo and kde
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

```python
#matrix plot->heatmap->cluster map
temp_df=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(temp_df,annot=True)
plt.figure(figsize=[15,10])
```

```python
temp_df3=gap[gap['continent']=='Asia'].pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(temp_df3,annot=True)
plt.figure(figsize=[100,100])
```
