```python
#from a pandas dataframe
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman

plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
#plotting multiple columns
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('ROHIT SHARMA VS VIRAT KOHLI')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='black',linestyle='solid',linewidth=2,marker='8')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='solid',linewidth=2,marker='8')

plt.title('ROHIT SHARMA VS VIRAT KOHLI')
plt.xlabel('Season')
plt.ylabel('Runs scored')
```

```python
#legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='black',linestyle='solid',linewidth=2,marker='8',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='solid',linewidth=2,marker='8',label='Rohit')

plt.title('ROHIT SHARMA VS VIRAT KOHLI')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='black',linestyle='solid',linewidth=2,marker='8',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='solid',linewidth=2,marker='8',label='Rohit')

plt.title('ROHIT SHARMA VS VIRAT KOHLI')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()
plt.grid()
```

```python
#show
plt.plot(batsman['index'],batsman['V Kohli'],color='black',linestyle='solid',linewidth=2,marker='8',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='solid',linewidth=2,marker='8',label='Rohit')

plt.title('ROHIT SHARMA VS VIRAT KOHLI')
plt.xlabel('Season')
plt.ylabel('Runs scored')
plt.legend()

plt.show() #need to call this function to view the plotted graph
```

```python
#for both bivariate and univariate analysis
```

```python
children =[50,30,40,50]
colors=['blue','red','green','yellow']
plt.bar(colors,children,color='pink')
```

```python
#horizontal bars
plt.barh(colors,children,color='pink')
```

```python
#color and label
df=pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.27,df['2015'],width=0.25)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.25)
plt.bar(np.arange(df.shape[0])+0.27,df['2017'],width=0.25)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
```

```python
#stacked bar chart
plt.bar(df['batsman'],df['2015'],bottom=df['2016']+df['2017'])
plt.bar(df['batsman'],df['2016'],bottom=df['2017'])
plt.bar(df['batsman'],df['2017'])
```
